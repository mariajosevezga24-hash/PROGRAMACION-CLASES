```python
import re

# Excepción personalizada
class CorreoInvalidoError(Exception):
    pass

# Función de validación
def validar_correo(correo):

    patron = r"^[a-z]{3,12}+_\d+_[a-z]{3,30}+@universitaria\.edu\.co$"
    if not re.fullmatch(patron, correo):
        raise CorreoInvalidoError(
            "Correo inválido. Use el formato: nombre_edad_apellido@universitaria.edu.co"
        )
    return True

# Programa principal
INTENTOS = 3


while 0 < INTENTOS:
    try:
        correo = input("Ingrese su correo institucional: ")
        validar_correo(correo)
        print("Correo válido")
        break
    except CorreoInvalidoError as e:
        INTENTOS -= 1
        print("Error:", e)
        if 0 < INTENTOS:
            print(f"Intentos restantes: {INTENTOS}")
    except Exception as e:
        print("Error inesperado:", e)
        break

if INTENTOS == 0:
    print("Acceso bloqueado por intentos fallidos.")
```

**Explicación del programa**

**Excepción personalizada**
Lo primero que se hizo fue crear una excepción personalizada llamada CorreoInvalidoError. Se decidió hacerla porque así el programa puede diferenciar cuándo el error es específicamente por un correo mal escrito y no por otra cosa. Para crearla simplemente se definió una clase que hereda de Exception y se le puso pass porque no necesita nada adicional.

**Función de validación**
Después se creó la función validar_correo que es la que hace todo el trabajo de verificar el correo. Dentro de ella se definió un patrón con expresión regular que describe exactamente cómo debe verse un correo válido: primero el nombre en minúsculas, luego un guión bajo, luego números, otro guión bajo, el apellido y finalmente el dominio @universitaria.edu.co. Se usó re.fullmatch() para que compare el correo completo contra ese patrón. Si no coincide, la función lanza la excepción personalizada que se creó antes con un mensaje explicando el formato correcto. Si sí coincide, retorna True.

**Constante de intentos**
Antes de entrar al bucle principal se definió la variable INTENTOS con valor 3, que representa cuántas veces el usuario puede equivocarse. Se escribió en mayúsculas siguiendo la convención de Python para indicar que es un valor fijo del programa.

**Bucle principal**
El núcleo del programa es un bucle while que se repite mientras queden intentos. Dentro se usó un bloque try/except para manejar los posibles errores. En el try se le pide el correo al usuario y se llama la función de validación. Si todo sale bien, se imprime que el correo es válido y se sale del bucle con break. Si la función lanza un CorreoInvalidoError, el except lo captura, descuenta un intento, muestra el mensaje de error y le avisa al usuario cuántos intentos le quedan. También se agregó un segundo except para capturar cualquier error inesperado que pueda ocurrir, en ese caso el programa termina directamente.

**Bloqueo final**
Por último, fuera del bucle se puso una condición que revisa si INTENTOS llegó a cero. Si fue así, significa que el usuario falló las tres veces y se muestra el mensaje de acceso bloqueado.
